Methods Used to Identify Missing and Unusual Values

Four methods were used to check missing and unusual values in the dataset:

1:unique() – Displays every unique value in a column, including NaN and indicators such as error, unknown, not available, or other 
inconsistent entries. It is particularly useful during data cleaning because all different types of values can be viewed at once.

2:value_counts(dropna=False) – Shows each unique value along with its frequency, including missing values. This is useful for understanding
how frequently a particular value or missing indicator occurs.

3:isnull() / isna() – Directly identifies actual NaN values and can be combined with .sum() to count them. This is the most direct
method for checking the number of missing values.

4:describe() – Provides a statistical summary of numerical columns, including the non-null count, mean, minimum, maximum, and quartiles.
It is useful for understanding the overall distribution and detecting potential unusual values.

Comparison: Among these methods, value_counts(dropna=false) was particularly useful for the initial data-cleaning stage because it provides a complete 
view of all distinct entries in one step. This makes it easier to identify not only NaN values but also text-based missing-value indicators 
such as error, unknown, and not available, which may not be detected by isnull() alone.

```python
#Checking the Mean Age

#The mean age was calculated before imputing the missing values in the age column to ensure that the replacement value was realistic. 
#The average was calculated using df['age'].mean(), which returned 38.45. This value represents a reasonable age and was therefore
#considered suitable for replacing the missing age values using mean imputation.

#Why Mean?

#Missing values need to be imputed with mean because the skewness value (0.0206) is close to zero, indicating an approximately symmetrical distribution.
#Which is very close to 0, there is no strong skewness pulling the mean away from the center.
```

```python
#Imputation with Mean,Mode,Median ,ffill() and bfill()

# Numerical columns with approximately symmetrical distribution
df['age'] = df['age'].fillna(round(df['age'].mean()))

df['likes'] = df['likes'].fillna(round(df['likes'].mean()))

df['comments'] = df['comments'].fillna(round(df['comments'].mean()))

# Categorical columns
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])

df['sentiment'] = df['sentiment'].fillna(df['sentiment'].mode()[0])

# Shares using forward fill and backward fill
df['shares'] = df['shares'].ffill().bfill()
```

Imputation with Mean,Mode,Median ,ffill() and bfill()

Missing values were handled based on the data type and distribution of each column. Mean imputation was used for age, likes, and 
comments because their skewness values were close to zero, indicating approximately symmetrical distributions.

Mode imputation was used for the categorical columns gender and sentiment because categorical values cannot be replaced using mean or median.

The shares column was filled using forward fill and backward fill to use nearby available values to match the activity of social media.

Avoided using Median imputation because this method is not necessary for these numerical columns. because their distributions were 
approximately symmetrical and mean imputation was suitable.

```python
#Creating Engagement Score: user-defined function 

def engagement_score(rate):
    if rate >= 0.964356:
        return 'High'
    else:
        return 'Low'
```

```python
#Creating Engagement Score:Using for loop 

engagement_scores = []

for rate in df['engagement_rate']:
    score = engagement_score(rate)
    engagement_scores.append(score)

df['engagement_score'] = engagement_scores
```

If the engagement rate is greater than or equal to the dataset average of 0.964356, it returns "High"; otherwise, it returns "Low".

Next, a for loop was used to go through each value in the "engagement_rate" column. The user-defined function was called for each value, and the resulting engagement score was stored in a list. Finally, the list was added to the DataFrame as a new column named "engagement_score".

This step demonstrates the use of a user-defined function, for loop, conditional statement, and comparison logic in the analysis.

```python
#Task 5 — Statistical Analysis 
#Compute descriptive stats for likes, comments, shares, watch_time, engagement_rate, followers:
#Mean, median, mode
#Standard deviation, variance
#Percentiles
#(Optional) Skewness and kurtosis

numeric_cols = [
    'likes',
    'comments',
    'shares',
    'watch_time_sec',
    'engagement_rate',
    'follower_count'
]

# Calculate descriptive statistics
means = df[numeric_cols].mean()
medians = df[numeric_cols].median()
modes = df[numeric_cols].mode().iloc[0]
stds = df[numeric_cols].std()
variances = df[numeric_cols].var()
skewness = df[numeric_cols].skew()
kurtosis = df[numeric_cols].kurt()

# Display statistics using zip() and for loop
for column, mean, median, mode, std, var, skew, kurt in zip(
    numeric_cols,
    means,
    medians,
    modes,
    stds,
    variances,
    skewness,
    kurtosis
):
    print("\nColumn:", column)
    print("Mean:", mean)
    print("Median:", median)
    print("Mode:", mode)
    print("Standard Deviation:", std)
    print("Variance:", var)
    print("Skewness:", skew)
    print("Kurtosis:", kurt)

# Percentiles
print("\nPercentiles:")
print(df[numeric_cols].quantile([0.25, 0.50, 0.75]))
```

Use of "zip()" Function

The "zip()" function was used to organize the descriptive statistics for the six numerical columns: "likes", "comments", "shares", "watch_time", "engagement_rate", and "followers".

The statistical values for mean, median, mode, standard deviation, variance, skewness, and kurtosis were calculated separately. The "zip()" function then pairs the corresponding results with each column name. A "for" loop was used to go through these paired values and display all the statistics together for each column.

This makes the output more structured, readable, and easy to compare. Instead of displaying each statistical measure separately for all columns, the results for each column are presented together.

For example, the output is organized as:

Column: likes

- Mean
- Median
- Mode
- Standard Deviation
- Variance
- Skewness
- Kurtosis

Then the same set of statistics is displayed for "comments", "shares", "watch_time", "engagement_rate", and "followers".

Therefore, "zip()" is helpful here because it connects related results and allows the "for" loop to display them in an organized column-wise format.

```python
#Plotly (Interactive)
#Interactive line chart/bar chart/bubble/scatter chart


import plotly.express as px

# Make sure posted_at is datetime
df['posted_at'] = pd.to_datetime(df['posted_at'])

# Calculate average engagement rate for each hour
hourly_engagement = (
    df.groupby(df['posted_at'].dt.hour)['engagement_rate']
    .mean()
    .reset_index()
)

hourly_engagement.columns = ['hour', 'average_engagement']

# Convert hour to AM/PM format for display
hourly_engagement['time'] = pd.to_datetime(
    hourly_engagement['hour'],
    format='%H'
).dt.strftime('%I %p')

# Create interactive bar chart
fig = px.bar(
    hourly_engagement,
    x='time',
    y='average_engagement',
    title='Peak Engagement Time',
    labels={
        'time': 'Posting Time',
        'average_engagement': 'Average Engagement Rate'
    }
)

fig.show()
```

Peak Engagement Time Analysis

The posted_at column was converted to datetime format, and the hour was extracted to analyze social media engagement at different posting times. The average engagement_rate was calculated for each hour and converted into AM/PM format for easier interpretation. An interactive bar chart was created using Plotly to visualize the average engagement rate by posting time. The analysis identified 12:00 AM (midnight) as the peak engagement time, with an average engagement rate of 0.96458. This indicates that posts published around midnight achieved the highest average engagement rate in the analyzed dataset.

Syntax Explanation

pd.to_datetime() converts the posted_at column into datetime format. The .dt.hour function extracts the hour from each posting time, while groupby() groups the records according to the extracted hour. The mean() function calculates the average engagement rate for each hour, and reset_index() converts the grouped result back into a DataFrame. The hour values are then formatted into AM/PM notation using pd.to_datetime() and .dt.strftime(). Finally, px.bar() from Plotly Express creates an interactive bar chart showing the average engagement rate for each posting time.
